# Import Dataset:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

data = pd.read_csv("C:/Users/Admin/Downloads/Automobile_data.csv")

data.head()
```

# Check missing values:

```python
data.isnull().sum()
```

```python
(data == '?').sum()
```

# Data Cleaning:

```python
data.replace('?', np.nan, inplace=True)
```

```python
data
```

# Numerical & Categorical Data:

```python
data['normalized-losses'] = pd.to_numeric(
    data['normalized-losses'], errors='coerce'
)

data['horsepower'] = pd.to_numeric(
    data['horsepower'], errors='coerce'
)

data['price'] = pd.to_numeric(
    data['price'], errors='coerce'
)
```

```python
data
```

```python
data.isnull().sum()
```

```python
num_cols = data.select_dtypes(include=np.number).columns

for col in num_cols:
    data[col].fillna(data[col].median(), inplace=True)
```

```python
data.isnull().sum()
```

```python
data
```

# Skewness Analysis:

```python
data.skew(numeric_only=True)
```

```python
data.hist(figsize=(15, 12))
plt.tight_layout()
plt.show()
```

```python
data['price_log'] = np.log1p(data['price'])
```

# Data Scaling:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numeric_features = data.select_dtypes(include=np.number).columns

data[numeric_features] = scaler.fit_transform(
    data[numeric_features]
)
```

# Final Dataset:

```python
data.head()
data.shape
data.info()
data.describe()
```

```python
X = data.drop('price', axis=1)
y = data['price']

print("X shape:", X.shape)
print("Y shape:", y.shape)
```

# EDA Visualizations:

# Univariate:

# 1. Histogram — Price:

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.histplot(data=data, x='price', bins=20, kde=True)

plt.title('Distribution of Car Prices')
plt.xlabel('Price')
plt.ylabel('Number of Cars')
plt.show()
```

# 2. Box Plot — Price:

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=data, x='price')


plt.title('Box Plot of Car Prices')
plt.xlabel('Price')
plt.show()
```

# 3. Pie Chart — Fuel Type:

```python

fuel_counts = data['fuel-type'].value_counts()


plt.figure(figsize=(7, 7))
plt.pie(
    fuel_counts,
    labels=fuel_counts.index,
    autopct='%1.1f%%',
    startangle=90
)


plt.title('Fuel Type Distribution')
plt.show()
```

# 4. Count Plot — Car Make:

```python

plt.figure(figsize=(12, 6))
sns.countplot(data=data, x='make')


plt.title('Number of Cars by Make')
plt.xlabel('Car Make')
plt.ylabel('Number of Cars')
plt.xticks(rotation=90)
plt.show()
```

# Bivariate Analysis:

```python
engine_price = data.groupby("engine-size")["price"].mean().reset_index()

plt.figure(figsize=(10, 5))
plt.plot(
    engine_price["engine-size"],
    engine_price["price"],
    marker="o"
)

plt.title("Average Price by Engine Size")
plt.xlabel("Engine Size")
plt.ylabel("Average Price")
plt.grid(True)
plt.show()
```

# Bar Plot — Average Price by Make:

```python

make_price = (
    data.groupby("make")["price"]
    .mean()
    .sort_values(ascending=False)
)


plt.figure(figsize=(12, 6))
make_price.plot(kind="bar")


plt.title("Average Car Price by Make")
plt.xlabel("Car Make")
plt.ylabel("Average Price")
plt.xticks(rotation=90)
plt.show()
```

# Box Plot — Price by Body Style:

```python

plt.figure(figsize=(10, 6))


sns.boxplot(
    data=data,
    x="body-style",
    y="price"
)


plt.title("Car Price by Body Style")
plt.xlabel("Body Style")
plt.ylabel("Price")
plt.show()
```

# Scatter Plot — Horsepower vs Price:

```python

plt.figure(figsize=(8, 6))


plt.scatter(
    data["horsepower"],
    data["price"]
)


plt.title("Horsepower vs Car Price")
plt.xlabel("Horsepower")
plt.ylabel("Price")
plt.show()
```

# Violin Plot — Price by Fuel Type:

```python

plt.figure(figsize=(8, 6))


sns.violinplot(
    data=data,
    x="fuel-type",
    y="price"
)


plt.title("Price Distribution by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Price")
plt.show()
```

# Relplot — Engine Size vs Price by Fuel Type:

```python

sns.relplot(
    data=data,
    x="engine-size",
    y="price",
    hue="fuel-type",
    kind="scatter",
    height=6,
    aspect=1.4
)


plt.title("Engine Size vs Price by Fuel Type")
plt.show()
```

# Joint Plot — Horsepower vs Price:

```python

sns.jointplot(
    data=data,
    x="horsepower",
    y="price",
    kind="scatter",
    height=7
)


plt.show()
```

# Crosstab — Fuel Type vs Body Style:

```python

fuel_body = pd.crosstab(
    data["fuel-type"],
    data["body-style"]
)


print(fuel_body)
```

```python
plt.figure(figsize=(10, 5))

fuel_body.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Fuel Type vs Body Style")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
plt.xticks(rotation=0)
plt.show()
```

# Multivariate Analysis:

# Pairplot:

```python
pair_cols = [
    "engine-size",
    "horsepower",
    "city-mpg",
    "highway-mpg",
    "price"
]

sns.pairplot(
    data[pair_cols],
    diag_kind="hist"
)

plt.show()
```

# Correlation Heatmap:

```python

numeric_cols = [
    "symboling",
    "normalized-losses",
    "engine-size",
    "horsepower",
    "city-mpg",
    "highway-mpg",
    "price"
]

correlation = data[numeric_cols].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Automobile Features")
plt.show()
```
